```python
import cv2
from google.colab.patches import cv2_imshow
```

```python
# Carregar imagem
imagem = cv2.imread('/content/drive/MyDrive/img_Rachaduras/image1.png') # , cv2.IMREAD_GRAYSCALE (Aplica preto e branco)
cv2_imshow(imagem)
```

```python
imagem.shape
```

```python
imagem = cv2.resize(imagem, (1040, 780))
cv2_imshow(imagem)
image_gray = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY) # Transforma a imagem em preto e branco
```

```python
# Aplicar filtro para remoção de ruído
imagem_filtrada = cv2.GaussianBlur(image_gray, (5, 5), 0)
# Detectar bordas usando Canny
bordas = cv2.Canny(imagem_filtrada, 50, 150)
# Mostrar resultado
cv2_imshow(bordas)
```
